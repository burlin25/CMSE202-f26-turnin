&#9989; **Question 3.2 (5 points):** With this data, we can also plot an approximate path that each tornado took.  For this part, do the following:
* Start with the same folium `Map` you used in Question 3.1
* Add a line for each tornado in MI that has the appropriate starting and ending coordinates - which folium command will you need to use here?  (2 points)
* You will find that some of these paths don't make sense. Find a way to filter or manipulate the data appropriately so that the paths are reasonable.  Your decision for how you handle the data will involve some trade-offs; you will justify your approach in the next problem. (2 points)
* Change the color to anything other than the default (1 point)
  
If all goes well, you should end up with something that looks like the map shown on the tornado tracker at the [Midwestern Regional Climate Center at Purdue](https://mrcc.purdue.edu/gismaps/cntytorn).

````python
# AI assistance: Codex helped inspect invalid coordinates and draft this filter/map.
# Folium PolyLine reference: https://python-visualization.github.io/folium/latest/reference.html#folium.vector_layers.PolyLine
valid_paths = mi_tornadoes[
    mi_tornadoes[["slat", "slon", "elat", "elon"]].notna().all(axis=1)
    & (mi_tornadoes["slat"] != 0)
    & (mi_tornadoes["slon"] != 0)
    & (mi_tornadoes["elat"] != 0)
    & (mi_tornadoes["elon"] != 0)
]

mi_path_map = folium.Map(location=[44.3, -85.6], zoom_start=6)
for _, row in valid_paths.iterrows():
    folium.PolyLine(
        locations=[[row["slat"], row["slon"]], [row["elat"], row["elon"]]],
        color="darkred",
        weight=2,
        opacity=0.6,
    ).add_to(mi_path_map)

print(f"Mapped {len(valid_paths)} paths; filtered out {len(mi_tornadoes) - len(valid_paths)} records.")
mi_path_map
````

````python
# AI assistance: Codex helped draft and verify this class against the supplied tests.
class VendingMachine:
    def __init__(self, starting_treats):
        self.inventory = starting_treats
        self.stash = 0
        self.current_transaction = 0

    def Restock(self, new_list):
        for item in new_list:
            self.inventory.append(item)

    def IsAvailable(self, item):
        return item in self.inventory

    def InputCash(self, amount):
        self.current_transaction += amount

    def Vend(self, item):
        self.inventory.remove(item)

    def MakeRequest(self, item):
        if self.current_transaction >= 1 and self.IsAvailable(item):
            self.Vend(item)
            self.stash += 1
            self.current_transaction -= 1
````
